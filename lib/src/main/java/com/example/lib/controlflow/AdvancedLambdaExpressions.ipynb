```kotlin
val originalText = "I don't know... what to say...123456"
originalText.filter {
    val isNotDigit = !it.isDigit()
    val stringRepresentation = it.toString()

    isNotDigit || stringRepresentation.toInt() >= 5
}
```

```kotlin
val originalText = "I don't know... what to say...123456"
originalText.filter {
    if (!it.isDigit()) {
        return@filter true
    }

    it.toString().toInt() >= 5
}
```

```kotlin
fun placeArgument(
    value: Int,
    f: (Int, Int) -> Int
) = { i: Int -> f(i, value) }

placeArgument(
    value =  2,
    f={ a, b -> a / b }
).invoke(4)
```

```kotlin
fun different(a: Int, b: Int) = a - b
val mul2 = { a: Int, b: Int -> a * b }

placeArgument(4, ::different).invoke(1)
placeArgument(3, mul2).invoke(3)
```

```kotlin
@Suppress("unused")
fun applyTransformation(
    numbers: List<Int>,
    transformation: (Int) -> Int
)=numbers.map { transformation(it) }
```

```kotlin
fun bigFunction(
    i:Int,
    f:(Int,Int)->Int
):(Int)->Int={a->f(a,i)}

bigFunction(1,{a,b->a+b})(1)
```

```kotlin
@Suppress("unused")
fun filterLongStrings(
    strings: List<String>,
    length: Int
)=strings.filter { it.length>=length }
```

```kotlin
private val peaceWords = arrayOf<Any>("I", "&&n", 3, "love", 1, "peace", "&&n")
private val pattern = Regex("\\d+")
peaceWords.filterNot {
    if (pattern.matches(it.toString())) {
        return@filterNot true
    }
    it.toString() == "&&n"
}
```
