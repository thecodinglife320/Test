```kotlin
import java.util.*

val listOfNames = listOf("John", "Jane", "Mary", "Peter", "Paul", "George")
val stackOfNames = Stack<String>()

stackOfNames.addAll(listOfNames)
while (stackOfNames.isNotEmpty()) {
    print(stackOfNames.pop())
    print(" ")
}
```

```kotlin
import java.util.Stack

val list = readln().split(" ")
val res = Stack<String>()
list.forEach { res.push(it) }
res.reversed().joinToString(
    separator = ", ",
    prefix = "[",
    postfix = "]",
    transform = {it}
).also(::println)
```

```kotlin
val list = readln().split(" ")
Stack<String>().run {
    addAll(list)
    val stringBuilder = StringBuilder()
    while (isNotEmpty()){ stringBuilder.append(pop()+" ")}
    stringBuilder.trim()
}
```

```kotlin
fun checkPalindrome(word: String): Boolean {
    return Stack<Char>().run {->
        addAll(word.asSequence())
        buildString {
            while (this@run.isNotEmpty()){
                append(pop())
            }
        }==word
    }
}

val input = readln()
println(checkPalindrome(input))
```

```kotlin
val list = readln().split(" ").map{ it.toDouble() }
print(Stack<Double>().run {
    addAll(list)
    buildString {
        while (this@run.isNotEmpty()){
            append("${pop()} ")
        }
    }.trim()
})
```

```kotlin
class Ship(val name: String, val ammunition: Int) {
    override fun toString(): String {
        return "$name-$ammunition"
    }
}

val ships = readln().split(" ")
val shipsList = ships.map { Ship(it.split("-")[0], it.split("-")[1].toInt()) }
Stack<Ship>().apply {
    addAll(shipsList)
    val ships = mutableListOf<Ship>()
    while (isNotEmpty()) {
        ships.add(pop())
    }
    ships.also(::print)
}
```
