```kotlin
val names = listOf("John", "Jane", "Mary", "Peter", "John", "Jane", "Mary", "Peter")

names.groupingBy { it.first() }.eachCount()

names
    .groupingBy { it.first() }
    .fold("") { acc, name -> acc + name }

names
    .groupingBy { it.length }
    .reduce { _, acc, name ->
        if (name.length > acc.length) name else acc
    }
```

```kotlin
val names = listOf(
    "John",
    "Jane",
    "Mary",
    "Peter",
    "John",
    "Jane",
    "Mary",
    "Peter"
)

names
    .groupingBy { it.first() }
    .aggregate { _, accumulator: Int?, _, first ->
        if (first) 1 else accumulator!! + 1
    }

// Grouping and using aggregate, returning even or odd size according to the size of the group
names
    .groupingBy { it.first() }
    .aggregate { _, accumulator: Int?, element, first ->
        when {
            first -> if (element.length % 2 == 0) element.length else 0
            element.length % 2 == 0 -> accumulator!! + element.length
            else -> accumulator
        }
    }
```
