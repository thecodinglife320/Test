```kotlin
ArrayDeque<Int>().apply {

    // as a queue, FIFO on both sides
    addLast(1)
    addLast(2)
    addLast(3)
    removeFirst()
    removeFirst()

    // as a stack, LIFO on one side, i.e., the end
    addLast(1)
    addLast(2)
    removeLast()
    removeLast()

    // or LIFO on the other side, i.e., the start
    addFirst(1)
    addFirst(2)
    removeFirst()
    removeFirst()
}
```

```kotlin
data class Ship(val name: String, val ammunition: Int) {
    override fun toString(): String {
        return "$name-$ammunition"
    }
}

fun fixTheQueue(input: String): ArrayDeque<Ship> {
    // Convert input to list of ships
    val shipsList = input.split(" ").map {
        val parts = it.split("-")
        Ship(parts[0], parts[1].toInt())
    }

    // Use ArrayDeque to reorganize a list
    // Put your code here

    return ArrayDeque<Ship>().apply { addAll(shipsList.reversed()) }
}
```

```kotlin
fun isPalindrome(word: String): Boolean {
    val deque = ArrayDeque<Char>()

    for (char in word) {
        deque.addLast(char)
    }

    return buildString {
        while (deque.isNotEmpty()) append(deque.removeLast())
    }==word
}

isPalindrome("refer")
```

```kotlin
fun printGradesFIFO(grades: ArrayDeque<Double>) {
    grades
        .joinToString(separator = " ") { it.toString() }
        .also(::print)
}

printGradesFIFO(ArrayDeque<Double>().apply {
    addAll(listOf(1.0,2.0))
})
```

```kotlin
fun processNamesLIFO(names: ArrayDeque<String>): ArrayDeque<String> {
    val outputStack = ArrayDeque<String>()
    return outputStack.apply {
        while (names.isNotEmpty()) outputStack.addLast(names.removeLast())
    }
}
processNamesLIFO(ArrayDeque<String>().apply {
    addAll("john jane joe jill pack".split(" "))
})
```
