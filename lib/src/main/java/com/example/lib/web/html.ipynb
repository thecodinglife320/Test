```kotlin
import kotlinx.html.*
import kotlinx.html.dom.*

```

```kotlin
document {
    append {
        html {
            head {
                title { +"My html page" }
            }
            body {
                h1 {
                    +"Hello use!"
                }
            }
        }
    }
}.serialize()
```

```kotlin
val comments = mutableListOf("Hello, everyone!", "Great site", "Nice :)")
document {
    append {
        html {
            body {
                for (comment in comments) {
                    div {
                        +comment
                    }
                }
            }
        }
    }
}.serialize()
```

```kotlin
document {
    append {
        html {
            body {
                a(href = "hyperskill.org", classes = "mylink") {
                    +"visit us"
                }
            }
        }
    }
}.serialize()
```

```kotlin
document {
    append {
        html {
            body {
                div {
                    apply { id = "mainDiv" }
                }
            }
        }
    }
    getElementById("mainDiv").append {
        div {
            apply {
                i {
                    apply {
                        id = "content"
                    }
                    +"dat dep trai"
                }
            }
        }
    }
//        .appendChild(createElement("div").apply {
//            append {
//                i { +"Hyperskill" }
//            }
//        })
}.serialize()
```

```kotlin
val myDocument = document { }
myDocument.append {
    html {
        body {
            h1 { +"Hello world!" }
            div {
                p { +"Paragraph 1" }
                p { +"Paragraph 2" }
            }
            p { +"Paragraph 3" }
        }
    }
}
println(myDocument.serialize())
```

```kotlin
val myDocument = document { }
myDocument.append {
    html {
        body {
            for (i in 1..10) {
                div {
                    +"$i"
                }
            }
        }
    }
}
println(myDocument.serialize())
```
